## Creating the full workflow

```python
import requests
```

```python
url1 = 'https://api.themoviedb.org/3/genre/movie/list?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US'
response1 = requests.get(url1)
data1 = response1.json()
```

```python
# creating a dictionary mapping genre_id with genre
genre_dict = {}

for dict in data1['genres']:
    id_num = dict['id']
    id_name = dict['name']

    genre_dict[id_num] = id_name
```

```python
genre_dict
```

```python
# from the description of the api we see that there are a total of 544 pages so to get info of all the 
# movies we need to change url everytime
# each URL has about 20 movies info
```

```python
# function that takes genre_id list as input and gives genres as output
def movie_genres(genre_id):
    new_list = []
    for id in genre_id:
        new_list.append(genre_dict[id])
    return new_list
```

```python
import csv
```

```python
def write_to_csv(writer, title, desc, genres):
    writer.writerow([title,desc,", ".join(genres)])
```

```python
with open('../Data/movies.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Title','Overview','Genres'])

    for i in range(1,501):
        # hitting the movie info API
        url2 = f'https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={i}'
        response2 = requests.get(url2)
        data2 = response2.json()

        if 'results' not in data2:
            print(f"Failed at page {i}")
            print(data2)
            continue

        for movie in data2['results']:
            # and every elements of this list it also a dictionary
            
            title = movie['original_title']
            desc = movie['overview']
            genres = movie_genres(movie['genre_ids'])

            write_to_csv(writer,title,desc,genres)
```
