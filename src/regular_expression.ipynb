```python
# ============================================================
#          REGULAR EXPRESSIONS IN PYTHON — MASTERCLASS
#          For Interview Preparation
# ============================================================
#
# WHAT IS A REGULAR EXPRESSION?
# A regular expression (regex / regexp) is a sequence of characters
# that defines a SEARCH PATTERN. It is not Python-specific — the same
# syntax works in JavaScript, Java, C++, grep, sed, and most languages.
#
# Python's regex engine is based on the NFA (Nondeterministic Finite
# Automaton) model, which means:
#   - It tries each alternative left-to-right.
#   - It backtracks when a path fails.
#   - This is powerful but can be slow on pathological patterns (see Section 11).
#
# WHY LEARN REGEX?
#   - String validation (email, phone, IP address, passwords)
#   - Text search and extraction (web scraping, log parsing)
#   - Find-and-replace operations
#   - Tokenisation in compilers and parsers
#   - Data cleaning in data science / ETL pipelines
#
# MODULE TO USE IN PYTHON:
import re
# The 're' module is part of Python's standard library — no installation needed.
```

## SECTION 1: RAW STRINGS — WHY YOU MUST USE THEM WITH REGEX

```python
# In a normal Python string, backslash has special meaning:
#   "\n" --> newline
#   "\t" --> tab
#   "\d" --> not a valid python escape sequence, but if we want to represent a literal backslash followed by 'd', 
#            we have to escape the backslash itself:
#
# In a RAW string (r"..."), backslashes are treated as literal characters.
#   r"\n" --> the two characters: \ and n
#   r"\d" --> the two characters: \ and d  (which regex engine reads as 'digit')
#
# RULE: Always prefix regex patterns with r"..." to avoid double-escaping.
 
normal  = "\\d+"   # you have to escape the backslash yourself
raw     = r"\d+"   # cleaner and less error-prone -- ALWAYS USE THIS
 
print(normal == raw)   # True -- same result, but raw is far more readable
```

## SECTION 2: THE re MODULE — ALL FUNCTIONS

```python
# ┌────────────────┬────────────────────────────────────────────────────┐
# │ Pattern        │ Meaning                                            │
# ├────────────────┼────────────────────────────────────────────────────┤
# │ .              │ Any char except \n                                 │
# │ \d / \D        │ Digit / Non-digit                                  │
# │ \w / \W        │ Word char / Non-word char                          │
# │ \s / \S        │ Whitespace / Non-whitespace                        │
# │ \b / \B        │ Word boundary / Non-boundary                       │
# │ \A / \Z        │ String start / String end (strict)                 │
# │ ^ / $          │ Line start / Line end                              │
# │ [abc]          │ One of a, b, c                                     │
# │ [^abc]         │ NOT one of a, b, c                                 │
# │ [a-z]          │ Range a to z                                       │
# │ *              │ 0 or more (greedy)                                 │
# │ +              │ 1 or more (greedy)                                 │
# │ ?              │ 0 or 1   (greedy)                                  │
# │ *? +? ??       │ Lazy versions of the above                         │
# │ {n}            │ Exactly n times                                    │
# │ {n,m}          │ Between n and m times (greedy)                     │
# │ {n,m}?         │ Between n and m times (lazy)                       │
# │ (...)          │ Capturing group                                    │
# │ (?:...)        │ Non-capturing group                                │
# │ (?P<name>...)  │ Named capturing group                              │
# │ (?P=name)      │ Backreference to named group                       │
# │ \1 \2 ...      │ Backreference to group 1, 2, ...                   │
# │ a|b            │ a OR b                                             │
# │ (?=...)        │ Positive lookahead                                 │
# │ (?!...)        │ Negative lookahead                                 │
# │ (?<=...)       │ Positive lookbehind (fixed width)                  │
# │ (?<!...)       │ Negative lookbehind (fixed width)                  │
# ├────────────────┼────────────────────────────────────────────────────┤
# │ Flag           │ Effect                                             │
# ├────────────────┼────────────────────────────────────────────────────┤
# │ re.I           │ Case-insensitive                                   │
# │ re.M           │ ^ $ match per line                                 │
# │ re.S           │ . matches \n too                                   │
# │ re.X           │ Verbose mode (allow whitespace + comments)         │
# ├────────────────┼────────────────────────────────────────────────────┤
# │ Function       │ Returns                                            │
# ├────────────────┼────────────────────────────────────────────────────┤
# │ re.match()     │ Match at START only                                │
# │ re.search()    │ First match ANYWHERE                               │
# │ re.fullmatch() │ Match must cover ENTIRE string                     │
# │ re.findall()   │ List of all matches (strings or tuples)            │
# │ re.finditer()  │ Iterator of Match objects                          │
# │ re.sub()       │ String with substitutions                          │
# │ re.subn()      │ (new_string, count) tuple                          │
# │ re.split()     │ List of split parts                                │
# │ re.compile()   │ Compiled Pattern object                            │
# ├────────────────┼────────────────────────────────────────────────────┤
# │ Match method   │ Returns                                            │
# ├────────────────┼────────────────────────────────────────────────────┤
# │ m.group()      │ Full matched string                                │
# │ m.group(n)     │ Content of group n                                 │
# │ m.groups()     │ Tuple of all captured groups                       │
# │ m.groupdict()  │ Dict of all named groups                           │
# │ m.start()      │ Start index of match                               │
# │ m.end()        │ End index of match                                 │
# │ m.span()       │ (start, end) tuple                                 │
# └────────────────┴────────────────────────────────────────────────────┘
#
# All functions return None (not an exception) when there is no match —
# EXCEPT findall/finditer which return empty list/iterator.
```

```python
# ── re.match() ──────────────────────────────────────────────
# Tries to match at the VERY BEGINNING of the string.
# Even if the pattern exists later, match() will NOT find it.
 
m = re.match(r"\d+", "123abc 345")
print(m.group())   # '123'
 
m = re.match(r"\d+", "abc123")
print(m)           # None  -- '123' is not at the start
```

```python
# ── re.search() ─────────────────────────────────────────────
# Scans the ENTIRE string and returns the FIRST match found anywhere.
 
m = re.search(r"\d+", "abc123def456")
print(m.group())   # 'abc123def456'  -- finds the first occurrence
```

```python
# ── re.fullmatch() ──────────────────────────────────────────
# The ENTIRE string must match the pattern — nothing more, nothing less.
# Very useful for input validation (e.g., "is this string a valid email?").
 
print(re.fullmatch(r"\d{4}", "2024"))    # Match object
print(re.fullmatch(r"\d{4}", "20245"))   # None -- extra character
```

```python
# ── re.findall() ────────────────────────────────────────────
# Returns a LIST of all non-overlapping matches.
# If the pattern has ONE capturing group  --> list of strings
# If the pattern has MULTIPLE groups      --> list of tuples
# If no groups                            --> list of full match strings
 
text = "price: 100, discount: 20, total: 80"
 
print(re.findall(r"\d+", text))
# ['100', '20', '80']
 
# With one capturing group:
print(re.findall(r"(\d+)", text))
# ['100', '20', '80']  -- same as above for single group
 
# With multiple groups:
print(re.findall(r"(\w+): (\d+)", text))
# [('price', '100'), ('discount', '20'), ('total', '80')]
```
